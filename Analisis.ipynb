### Transformación 2: convertir el precio a número

Creamos `PUBLICO_VENTA_NUM` a partir de `PUBLICO VENTA`, reemplazando la coma decimal por un punto para que pandas pueda interpretar los valores como números. Conservamos la columna original.

```python
df["PUBLICO_VENTA_NUM"] = pd.to_numeric(
    df["PUBLICO VENTA"].str.replace(",", ".", regex=False)
)

print(df[["PUBLICO VENTA", "PUBLICO_VENTA_NUM"]].head())
print("Tipo:", df["PUBLICO_VENTA_NUM"].dtype)
```

### Agrupar las fechas por mes

Creamos `MES` a partir de `FECHA_DT` para organizar los registros mensualmente. Conservamos las fechas originales y completas.

```python
df["MES"] = df["FECHA_DT"].dt.to_period("M").astype(str)
sorted(df["MES"].unique())
```

### Resumen mensual exploratorio de precios

Agrupamos los registros por mes para calcular la cantidad de registros, la media y la mediana del precio unitario. Este resumen es exploratorio porque la mezcla de productos puede cambiar entre meses.

```python
resumen_mensual = df.groupby("MES")["PUBLICO_VENTA_NUM"].agg(
    cantidad_registros="count",
    precio_medio="mean",
    precio_mediano="median"
)

resumen_mensual
```

### Presencia de productos por cantidad de meses

Contamos en cuántos meses aparece cada descripción de producto. Esto permite evaluar cuántos productos podríamos comparar a lo largo del período.

```python
meses_por_producto = df.groupby("DESCRIPX")["MES"].nunique()

meses_por_producto.value_counts().sort_index()
```

### Verificación del identificador de producto

Comprobamos la relación entre `DESCRIPX` y `TROQUEL` para elegir un identificador consistente al comparar productos entre meses.

```python
print(
    "Máximo de descripciones por TROQUEL:",
    df.groupby("TROQUEL")["DESCRIPX"].nunique().max()
)
print(
    "Máximo de TROQUEL por descripción:",
    df.groupby("DESCRIPX")["TROQUEL"].nunique().max()
)
```

### Productos presentes en todos los meses

Identificamos los códigos `TROQUEL` registrados en los 13 meses. Este conjunto permitirá comparar los mismos productos durante todo el período.

```python
meses_por_troquel = df.groupby("TROQUEL")["MES"].nunique()
troqueles_todos_los_meses = meses_por_troquel[
    meses_por_troquel == 13
].index

print("Productos presentes en los 13 meses:", len(troqueles_todos_los_meses))
```

### Precio promedio por producto y mes

Tomamos los 298 códigos `TROQUEL` presentes en todos los meses y calculamos el precio unitario promedio de cada producto dentro de cada mes. Así resumimos posibles ventas repetidas sin dar más peso a un producto por aparecer en más registros.

```python
df_comparacion = df[
    df["TROQUEL"].isin(troqueles_todos_los_meses)
].copy()

precio_producto_mes = (
    df_comparacion
    .groupby(["TROQUEL", "MES"])
    .agg(
        descripcion=("DESCRIPX", "first"),
        precio_promedio=("PUBLICO_VENTA_NUM", "mean")
    )
    .reset_index()
)

print("Filas del resumen:", len(precio_producto_mes))
display(precio_producto_mes.head())
```

### Organizar precios por producto y mes

Convertimos el resumen en una matriz con un producto por fila y un mes por columna. Verificamos que estén los 13 meses y que no falten precios.

```python
precios_matriz = precio_producto_mes.pivot(
    index="TROQUEL",
    columns="MES",
    values="precio_promedio"
)

print("Dimensiones:", precios_matriz.shape)
print("Valores faltantes:", precios_matriz.isna().sum().sum())
display(precios_matriz.iloc[:5, :5])
```

### Verificar precios del mes base

Comprobamos que los precios promedio de septiembre de 2025 sean mayores que cero antes de usarlos como base para comparar la evolución de cada producto.

```python
print(
    "Precios base iguales a cero:",
    (precios_matriz["2025-09"] == 0).sum()
)
```

### Índice de precio por producto

Expresamos el precio promedio mensual de cada producto como índice, tomando septiembre de 2025 como base 100. Esto permite comparar la evolución relativa de productos con precios distintos.

```python
indice_productos = (
    precios_matriz
    .div(precios_matriz["2025-09"], axis=0)
    * 100
)

display(indice_productos.iloc[:5, :5].round(2))
```

### Evolución promedio de los productos

Calculamos el promedio mensual de los índices individuales de los 298 productos. Cada producto tiene el mismo peso. También mostramos la variación porcentual respecto de septiembre de 2025.

```python
resumen_indice = pd.DataFrame({
    "indice_promedio": indice_productos.mean(axis=0)
})

resumen_indice["variacion_desde_base_%"] = (
    resumen_indice["indice_promedio"] - 100
)

display(resumen_indice.round(2))
```

### Comparación con la inflación

Comparamos la variación del índice promedio de los 298 productos entre septiembre de 2025 y septiembre de 2026 con la variación interanual del IPC nacional del INDEC para septiembre de 2026. La diferencia se expresa en puntos porcentuales.

Fuente de inflación: [INDEC, IPC de septiembre de 2026](https://www.indec.gob.ar/uploads/informesdeprensa/ipc_10_268C45A9444C.pdf).

```python
variacion_farmacia = resumen_indice.loc[
    "2026-09", "variacion_desde_base_%"
]

inflacion_ipc = 33.4

diferencia_pp = variacion_farmacia - inflacion_ipc

print(f"Variación de los productos: {variacion_farmacia:.2f}%")
print(f"Inflación IPC nacional: {inflacion_ipc:.2f}%")
print(f"Diferencia: {diferencia_pp:.2f} puntos porcentuales")
```

### Evolución mensual del índice de precios

Graficamos el índice promedio de los 298 productos, con septiembre de 2025 como base 100. La línea permite observar cómo evolucionó el conjunto seleccionado durante los 13 meses. El gráfico se guarda para incluirlo luego en el informe.

```python
plt.figure(figsize=(10, 5))
plt.plot(
    resumen_indice.index,
    resumen_indice["indice_promedio"],
    marker="o",
    color="darkgreen"
)
plt.title("Evolución del precio promedio de 298 medicamentos")
plt.xlabel("Mes")
plt.ylabel("Índice promedio (septiembre 2025 = 100)")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("evolucion_precios_farmacia.png", dpi=300, bbox_inches="tight")
plt.show()
```
